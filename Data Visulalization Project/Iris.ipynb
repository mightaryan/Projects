```python
# Importing Necessary Libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Step - 1 ==> Reading Data 
data = pd.read_csv("Iris.csv")
df = pd.DataFrame(data)

# df.head(15)
# df.info(), df.describe()

# Step - 2 ==> EDA - Pairplot
"""
Pairplot to visualize relationships between features colored by species.
pairplot shows scatter plots for each pair of features and histograms for each feature on the diagonal.
pairplot helps in understanding the distribution and relationships of features across different species.
"""
plt.figure(figsize=(8, 6))
sns.pairplot(df, hue="Species", markers=["o", "s", "D"], diag_kind="hist")
plt.suptitle("Pairplot of Iris Dataset", y=1.02)
sns.set_theme(style="whitegrid")
plt.show()

# Step - 3 ==> EDA - Correlation Heatmap
"""
Correlation heatmap to visualize relationships between numerical features.
The heatmap displays correlation coefficients between features, with colors indicating the strength and direction of correlations.
A high positive correlation (close to 1) indicates that as one feature increases, the other tends to increase as well, while a high negative correlation (close to -1) indicates an inverse relationship.
"""
plt.figure(figsize=(8, 6))
correlation_matrix = df.iloc[:, 1:5].corr()
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Iris Dataset")
plt.show()

# Step - 4 ==> EDA - Boxplot
"""Boxplot to visualize the distribution of features across different species.
Boxplots show the median, quartiles, and potential outliers for each feature, grouped by species.
This helps in comparing the spread and central tendency of features among the three species of Iris flowers"""
plt.figure(figsize=(10, 6))
df_melted = pd.melt(df, id_vars=["Species"], value_vars=["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"])
sns.boxplot(x="variable", y="value", hue="Species", data=df_melted)
plt.title("Boxplot of Iris Features by Species")
plt.show()

# Step - 5 ==> EDA - Violin Plot
"""Violin plot to visualize the distribution of petal length across different species.
Violin plots combine aspects of boxplots and kernel density plots, showing the distribution shape of the data.
This helps in understanding the density and spread of petal lengths for each species of Iris flowers"""
plt.figure(figsize=(10, 6))
sns.violinplot(x="Species", y="PetalLengthCm", data=df, palette="Set2")
plt.title("Violin Plot of Petal Length by Species")
plt.show()


# Step - 6 ==> EDA - Scatter Plot
"""Scatter plot to visualize the relationship between sepal length and sepal width, colored by species.
Scatter plots help in identifying patterns and clusters among different species based on sepal dimensions."""
plt.figure(figsize=(8, 6))
sns.scatterplot(x="SepalLengthCm", y="SepalWidthCm", hue="Species", style="Species", s=100, data=df)
plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.show()  

# Step - 7 ==> EDA - Histogram
"""Histograms to visualize the distribution of each feature in the Iris dataset.
Histograms show the frequency distribution of feature values, helping to identify patterns and differences among species."""
plt.figure(figsize=(10, 6))
df.iloc[:, 1:5].hist(bins=15, edgecolor='black', layout=(2, 2), figsize=(10, 8))
plt.suptitle("Histograms of Iris Features", y=1.02)
plt.show()


"""Conclusion:
# The exploratory data analysis (EDA) of the Iris dataset reveals distinct patterns and relationships among the features.
# The pairplot indicates clear separations between species based on petal and sepal dimensions. The correlation heatmap shows strong positive correlations between petal length and petal width, 
suggesting these features are closely related. Boxplots and violin plots highlight the variability and distribution of features across species, with
Iris-setosa generally having smaller dimensions compared to Iris-versicolor and Iris-virginica. Scatter plots further illustrate the clustering of species based on sepal dimensions.
# Histograms provide insights into the frequency distribution of each feature, confirming the differences in measurements among species. Overall,
the EDA suggests that the features are effective in distinguishing between the three Iris species and can be useful for classification tasks."""
```
